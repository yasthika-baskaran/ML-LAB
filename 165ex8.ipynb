```python
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.estimators import MaximumLikelihoodEstimator

# Define the structure of the Bayesian Network
# Removed the edge ('heartdisease', 'Lifestyle') to break the loop
model = DiscreteBayesianNetwork([
    ('age', 'heartdisease'),
    ('Gender', 'heartdisease'),
    ('Family', 'heartdisease'),
    ('diet', 'heartdisease'),
    ('Lifestyle', 'heartdisease'), # Keeping this edge, assuming Lifestyle influences heartdisease
    ('cholestrol', 'heartdisease')
])

# Fit the model using the MaximumLikelihoodEstimator
model.fit(df, estimator=MaximumLikelihoodEstimator)

# Print the Conditional Probability Distribution for 'heartdisease'
print(model.get_cpds('heartdisease'))
```
