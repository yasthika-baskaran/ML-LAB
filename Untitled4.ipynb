```python
# Credit Card Fraud Detection
# Using Random Forest Classifier

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

print("Libraries imported successfully!")

```

```python
try:
    # Running in Google Colab
    from google.colab import files

    print("Google Colab detected.")
    print("Please upload your creditcard.csv file:")

    uploaded = files.upload()
    filename = list(uploaded.keys())[0]

except ImportError:
    # Running in Jupyter Notebook / local Python
    print("Jupyter/local environment detected.")
    print("Make sure 'creditcard.csv' is in the same folder as this notebook.")

    filename = "creditcard.csv"

# Load dataset
df = pd.read_csv(filename)

print("\nDataset loaded successfully!")
print("Number of rows    :", df.shape[0])
print("Number of columns :", df.shape[1])
```

```python
print("First 5 rows of the dataset:")
display(df.head())
```

```python
print("Dataset Information:")
df.info()
```

```python
print("Missing values in each column:")
print(df.isnull().sum())

print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
print("Transaction Class Distribution:")
print(df["Approved"].value_counts())

print("\n0 = Not Approved")
print("1 = Approved")
```

```python
class_counts = df["Approved"].value_counts().sort_index()

plt.figure(figsize=(6, 4))

plt.bar(
    ["Not Approved", "Approved"],
    class_counts.values,
    color=["steelblue", "red"]
)

plt.title("Approved vs Not Approved Transactions")
plt.xlabel("Transaction Class")
plt.ylabel("Number of Transactions")

plt.show()
```

```python
# Input features
X = df.drop("Approved", axis=1)

# Target variable
y = df["Approved"]

print("Input features shape :", X.shape)
print("Target shape         :", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples :", X_train.shape[0])
print("Testing samples  :", X_test.shape[0])

print("\nTraining class distribution:")
print(y_train.value_counts())

print("\nTesting class distribution:")
print(y_test.value_counts())
```

```python
# Identify categorical columns
categorical_cols = X.select_dtypes(include='object').columns

# Apply one-hot encoding separately
X_train_encoded = pd.get_dummies(X_train, columns=categorical_cols, drop_first=True)
X_test_encoded = pd.get_dummies(X_test, columns=categorical_cols, drop_first=True)

# Align columns - crucial step
# Reindex X_test_encoded to match the columns of X_train_encoded, filling new columns with 0
missing_cols_in_test = set(X_train_encoded.columns) - set(X_test_encoded.columns)
for c in missing_cols_in_test:
    X_test_encoded[c] = 0
# Ensure the order of columns is the same
X_test_encoded = X_test_encoded[X_train_encoded.columns]

# Initialize StandardScaler
scaler = StandardScaler()

# Fit scaler ONLY on training data and transform
X_train_scaled = scaler.fit_transform(X_train_encoded)

# Transform test data using the same scaler
X_test_scaled = scaler.transform(X_test_encoded)

print("Feature scaling completed successfully!")
print("Training data shape:", X_train_scaled.shape)
print("Testing data shape :", X_test_scaled.shape)
```

```python
model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    class_weight="balanced",
    n_jobs=-1
)

print("Random Forest model created successfully!")
```

```python
print("Training the model...")

model.fit(X_train_scaled, y_train)

print("Model training completed successfully!")
```

```python
y_pred = model.predict(X_test_scaled)

print("Prediction completed!")

print("\nFirst 20 predictions:")
print(y_pred[:20])
```

```python
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, zero_division=0)
recall = recall_score(y_test, y_pred, zero_division=0)
f1 = f1_score(y_test, y_pred, zero_division=0)

print("MODEL PERFORMANCE")
print("-------------------------")
print(f"Accuracy  : {accuracy * 100:.2f}%")
print(f"Precision : {precision * 100:.2f}%")
print(f"Recall    : {recall * 100:.2f}%")
print(f"F1 Score  : {f1 * 100:.2f}%")
```

```python
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)

tn, fp, fn, tp = cm.ravel()

print("\nTrue Negatives  :", tn)
print("False Positives :", fp)
print("False Negatives :", fn)
print("True Positives  :", tp)
```

```python
plt.figure(figsize=(6, 5))

plt.imshow(cm, cmap="Blues")

plt.title("Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")

plt.xticks([0, 1], ["Genuine", "Fraud"])
plt.yticks([0, 1], ["Genuine", "Fraud"])

# Display values inside the matrix
for i in range(2):
    for j in range(2):
        plt.text(
            j,
            i,
            cm[i, j],
            ha="center",
            va="center",
            color="black",
            fontsize=12
        )

plt.colorbar()
plt.tight_layout()
plt.show()
```

```python
print("Classification Report:\n")

print(
    classification_report(
        y_test,
        y_pred,
        target_names=["Genuine", "Fraud"],
        zero_division=0
    )
)
```

```python
# Select one transaction from the test dataset
# Get the sample as a DataFrame to retain column names for encoding
sample_df = X_test.iloc[[0]].copy()

# Identify categorical columns (same as before)
categorical_cols = X.select_dtypes(include='object').columns

# Apply one-hot encoding to the sample
sample_encoded = pd.get_dummies(sample_df, columns=categorical_cols, drop_first=True)

# Align columns of the sample with the training data columns
# Add missing columns to the sample (if any) and fill with 0
missing_cols_in_sample = set(X_train_encoded.columns) - set(sample_encoded.columns)
for c in missing_cols_in_sample:
    sample_encoded[c] = 0
# Ensure the order of columns is the same
sample_encoded = sample_encoded[X_train_encoded.columns]

# Scale the encoded transaction using the scaler fitted on training data
sample_scaled = scaler.transform(sample_encoded)

# Make prediction
prediction = model.predict(sample_scaled)

if prediction[0] == 1:
    print("Approved Transaction")
else:
    print("Not Approved Transaction")
```
